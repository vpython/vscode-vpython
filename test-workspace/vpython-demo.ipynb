# VPython in VS Code — demo

Needs: the `feat/vscode-frontend` vpython-jupyter branch installed in this
kernel, and the VPython extension installed in VS Code. The scene appears
under **cell 1** (the import cell owns the scene container).

```python
from vpython import *
print('vpython imported — scene output above/below this line')
```

```python
floor = box(pos=vec(0, -1, 0), size=vec(6, 0.2, 6), color=color.green)
ball = sphere(pos=vec(0, 2, 0), radius=0.4, color=color.red, make_trail=True)
ball.v = vec(0.4, 0, 0.2)
g = vec(0, -9.8, 0)
print('objects created')
```

```python
# Bouncing ball — ~10 seconds of animation
dt = 0.01
for _ in range(1000):
    rate(100)
    ball.v = ball.v + g * dt
    ball.pos = ball.pos + ball.v * dt
    if ball.pos.y - ball.radius < floor.pos.y + 0.1:
        ball.v.y = -ball.v.y * 0.9
print('done')
```

```python
# Fonts + textures (needs extension >= 0.1.3)
sign = text(text='Hello VPython', align='center', height=0.6,
            color=color.yellow, pos=vec(0, 2.8, -1))
earth = sphere(pos=vec(-2, 1.5, -1), radius=0.6, texture=textures.earth)
print('text + textured sphere created')
```
